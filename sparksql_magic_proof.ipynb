# SparkSQL magic validation
This notebook validates `%%sql` from `jupyterlab-sparksql` against a Spark temp view.


```python
from pyspark.sql import SparkSession
import sparkmonitor
from pathlib import Path
```

```python
print(sparkmonitor.__path__ [0])
```

```python

listener_jar = Path(sparkmonitor.__path__[0]) / "listener_spark4_2.13.jar"
spark = (SparkSession.builder
    .config(
        "spark.extraListeners",
        "sparkmonitor.listener.JupyterSparkMonitorListener",
    )
    .config("spark.driver.extraClassPath", str(listener_jar))
    .appName("SparkSQLMagicJupyterLabProof")
    .getOrCreate())

df_filtered = spark.createDataFrame([
    ("notebook", 3, 42.5),
    ("spark", 7, 99.0),
    ("sql", 5, 64.25),
], ["item", "qty", "score"])
df_filtered.createOrReplaceTempView("df_filtered")
spark.version
```

```python
%%sql -n 10
select item, qty, score from df_filtered order by item
```

```python
spark.stop()
```
