```python
import pyspark
import sparkmonitor
from pyspark.sql import SparkSession
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

```python
print(sparkmonitor.__path__)
```

```python
from pathlib import Path

listener_jar = Path(sparkmonitor.__path__[0]) / "listener_spark4_2.13.jar"
spark_SparkMonitor = (SparkSession.builder
    .config(
        "spark.extraListeners",
        "sparkmonitor.listener.JupyterSparkMonitorListener",
    )
    .config("spark.driver.extraClassPath", str(listener_jar))
    .appName("PySpark-Get-Started")
    .getOrCreate()
)
```

```python
spark_SparkMonitor
```

```python
data = [
    (1, "a", "lisa@gmail.com"),
    (2, "b", "bob@gmail.com"),
    (3, "c", "carl@gmail.com"),
    (4, "d", "david@gmail.com"),
    (5, "e", "eve@gmail.com"),
    (6, None, "fred@gmail.com"),
    (7, "g", "gina@gmail.com"),
    (8, "h", None),
    (9, "i", "ian@gmail.com"),
]
schema = StructType(
    [
        StructField("id", IntegerType(), False),
        StructField("name", StringType(), True),
        StructField("email", StringType(), True),
    ]
)
df = spark_SparkMonitor.createDataFrame(data, schema)
df_filtered = df.filter((df.name.isNotNull()) & (col("id") >= 5))
```

```python
df_filtered.explain()
```

```python
df_filtered.printSchema()
```

```python
df_filtered.createOrReplaceTempView("df_filtered")
```

```python
%%sql -n 2
select * from df_filtered
```
